```python
import matplotlib.pyplot as plt
fruit_sale=[100,200,300,400]
month=["jan","feb","mar","apr"]
plt.figure(figsize=(8,10))
plt.bar( month,fruit_sale)
plt.bar(month, fruit_sale , color= ["purple","green","orange", "pink"],hatch='+')
plt.title("Title")
plt.xlabel("month")
plt.ylabel("amount")
plt.show()
```

```python
x=[1,2,56,4,5]
y=[10,30,90,40,50]
plt.plot(x,y)
```

```python
study_hours = [1,2,3,4,5]
exam_score=[55,60,70,85,90]
plt.scatter(study_hours, exam_score, color="black", alpha=0.9)
plt.title("scatter plot")
plt.xlabel("study_hour")
plt.ylabel("exam_score")
plt.show()
```

```python
labels=["apple", "samsung", "microsoft","google"]
scale =[20,25,30,10]
plt.figure(figsize=(10,5))
plt.pie(scale,labels=labels, autopct="%1.1f%%", shadow ="true")
plt.title("company and sales")
plt.show()
```

```python
months = ["jan", "feb", "mar", "apr", "may", "jun",
          "jul", "aug", "sep", "oct", "nov", "dec"]
sales_24=[50,53,56,45,67,89,23,45,23,78,90,11]
sales_25=[]
for i in sales_24:
  sales_25.append(i+10)
plt.plot(months,sales_24, label= "sales 2024", color ="blue", marker="o")
plt.plot(months,sales_25, label= "sales 2025", color ="green", marker="s")
plt.title("Title")
plt.xlabel("month")
plt.ylabel("amount")
plt.xticks(rotation=45, ha='right')
plt.show()
```
